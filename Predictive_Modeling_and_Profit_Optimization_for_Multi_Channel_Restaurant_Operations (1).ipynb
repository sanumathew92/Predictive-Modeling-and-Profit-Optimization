```python
# Define IQR Function

def find_outliers(column):

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5*IQR

    upper_limit = Q3 + 1.5*IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    return outliers
```

```python
# Detect Outliers in Delivery cost
outliers_delivery = find_outliers(
    "DeliveryCostPerOrder"
)

print(outliers_delivery.shape)
```

```python
# Detect Outliers in Commission Rate
outliers_commission = find_outliers(
    "CommissionRate"
)

print(outliers_commission.shape)
```

```python
# Detect Outliers in Profit Margin
outliers_profitmargin = find_outliers(
    "ProfitMargin"
)

print(outliers_profitmargin.shape)
```

# Winsorization (Treat Extreme values)
Instead of removing observations, cap them

```python
def winsorize(column):

  Q1 = df[column].quantile(0.25)

  Q3 = df[column].quantile(0.75)

  IQR = Q3 - Q1

  lower_limit = Q1 - 1.5*IQR

  upper_limit = Q3 + 1.5*IQR

  df[column] = np.where(
      df[column] < lower_limit,
      lower_limit,
      df[column]
  )

  df[column] = np.where(
      df[column] > upper_limit,
      upper_limit,
      df[column]
  )
```

```python
# Treat Delivery Cost Outliers
winsorize("DeliveryCostPerOrder")
```

```python
# Treat Commission Rate Outliers
winsorize("CommissionRate")
```

```python
# Treat Profit Margin Outliers
winsorize("ProfitMargin")
```

# Scipy Optimization

```python
from scipy.optimize import minimize

# Objective

feature_names = joblib.load(
    "Models/feature_names.pkl"
)

def objective(x):

    ue = x[0]
    dd = x[1]
    sd = x[2]

    temp = base_input.copy()

    temp["UE_share"] = ue
    temp["DD_share"] = dd
    temp["SD_share"] = sd

    # Update InStoreShare
    temp["InStoreShare"] = (
        1
        - ue
        - dd
        - sd
    )

    # Keep only model features
    temp = temp[feature_names]

    profit = best_model.predict(
        temp
    )[0]

    return -profit
```

```python
# Constraint

constraint={
'type':'eq',
'fun':lambda x:
x[0]+x[1]+x[2]-1
}
```

```python
# Bounds

bounds=[
(0,1),
(0,1),
(0,1)
]
```

```python
# Initial guess
x0=[.3,.3,.4]
```

```python
# Optimize

result=minimize(

objective,

x0,

bounds=bounds,

constraints=constraint
)
```

```python
# Optimal mix

optimal_mix=result.x

print(optimal_mix)
```

# Optimization Uplift

```python
optimized_profit=-result.fun

uplift_pct=(
optimized_profit-current_profit
)/current_profit*100

print(uplift_pct)
```
